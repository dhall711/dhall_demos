# Customer Data Schema Generator for Snowflake

This notebook creates synthetic customer data tables from various systems and a target customer schema with rich metadata for schema mapping exercises.

## Overview
- **Source Systems**: Salesforce CRM, Marketing Automation, Billing System, Support System
- **Target Schema**: Unified customer data table with semantic metadata
- **Metadata**: COMMENTs and TAGs for semantic understanding


## 1. Setup and Database Creation


```python
-- Create database and schema for our customer data mapping exercise
--CREATE OR REPLACE DATABASE CUSTOMER_DATA_MAPPING;
--USE DATABASE CUSTOMER_DATA_MAPPING;

CREATE OR REPLACE SCHEMA SOURCE_SYSTEMS;
CREATE OR REPLACE SCHEMA TARGET_SCHEMA;

-- Create tags for categorizing our data
CREATE OR REPLACE TAG DATA_MAPPER_DB.SOURCE_SYSTEMS.DATA_CLASSIFICATION;
CREATE OR REPLACE TAG DATA_MAPPER_DB.SOURCE_SYSTEMS.SYSTEM_SOURCE;
CREATE OR REPLACE TAG DATA_MAPPER_DB.SOURCE_SYSTEMS.DATA_SENSITIVITY;
CREATE OR REPLACE TAG DATA_MAPPER_DB.SOURCE_SYSTEMS.BUSINESS_DOMAIN;

-- Create tags for categorizing our data
CREATE OR REPLACE TAG DATA_MAPPER_DB.TARGET_SCHEMA.DATA_CLASSIFICATION;
CREATE OR REPLACE TAG DATA_MAPPER_DB.TARGET_SCHEMA.SYSTEM_SOURCE;
CREATE OR REPLACE TAG DATA_MAPPER_DB.TARGET_SCHEMA.DATA_SENSITIVITY;
CREATE OR REPLACE TAG DATA_MAPPER_DB.TARGET_SCHEMA.BUSINESS_DOMAIN;
```

## 2. Salesforce CRM System Table

Contains leads, contacts, and opportunity data from Salesforce Sales CRM


```python
USE SCHEMA SOURCE_SYSTEMS;

-- Salesforce CRM Contacts and Leads Table
CREATE OR REPLACE TABLE SF_CONTACTS_LEADS (
    SFDC_ID STRING COMMENT 'Salesforce unique identifier for contact/lead record',
    FIRST_NAME STRING COMMENT 'Given name of the contact or lead person',
    LAST_NAME STRING COMMENT 'Family name of the contact or lead person',
    EMAIL_ADDRESS STRING COMMENT 'Primary email address for communication',
    PHONE_NUMBER STRING COMMENT 'Primary phone contact number',
    COMPANY_NAME STRING COMMENT 'Organization or company name where person works',
    JOB_TITLE STRING COMMENT 'Professional role or position title',
    LEAD_STATUS STRING COMMENT 'Current status in sales pipeline (New, Qualified, Converted, etc.)',
    LEAD_SOURCE STRING COMMENT 'Original source of lead generation (Web, Referral, Event, etc.)',
    ANNUAL_REVENUE NUMBER COMMENT 'Company annual revenue in USD',
    EMPLOYEE_COUNT NUMBER COMMENT 'Total number of employees at company',
    INDUSTRY STRING COMMENT 'Business industry classification',
    STREET_ADDRESS STRING COMMENT 'Physical street address',
    CITY STRING COMMENT 'City name',
    STATE_PROVINCE STRING COMMENT 'State or province code',
    POSTAL_CODE STRING COMMENT 'ZIP or postal code',
    COUNTRY_CODE STRING COMMENT 'ISO country code',
    CREATED_DATE TIMESTAMP COMMENT 'Record creation timestamp in Salesforce',
    MODIFIED_DATE TIMESTAMP COMMENT 'Last modification timestamp',
    OWNER_ID STRING COMMENT 'Salesforce user ID of record owner'
)
COMMENT = 'Salesforce CRM contacts and leads data containing prospect and customer information'
TAG (SYSTEM_SOURCE = 'SALESFORCE_CRM', DATA_CLASSIFICATION = 'CUSTOMER_DATA', 
     DATA_SENSITIVITY = 'PII', BUSINESS_DOMAIN = 'SALES');
```

## 6. Target Customer Data Schema

This is the unified customer data table that we want to map all source systems to. Note that column names are intentionally different from source systems to demonstrate mapping requirements.


```python
USE SCHEMA TARGET_SCHEMA;

-- Unified Target Customer Data Table
CREATE OR REPLACE TABLE UNIFIED_CUSTOMER_PROFILE (
    -- Identity and Basic Information
    CUSTOMER_GUID STRING COMMENT 'Globally unique identifier for customer across all systems and touchpoints',
    INDIVIDUAL_GIVEN_NAME STRING COMMENT 'Personal first name or given name of the individual contact person',
    INDIVIDUAL_FAMILY_NAME STRING COMMENT 'Personal surname or family name of the individual contact person',
    INDIVIDUAL_FULL_NAME STRING COMMENT 'Complete formatted name combining first, middle, and last name components',
    PRIMARY_EMAIL_ADDRESS STRING COMMENT 'Primary email address for all business communications and system notifications',
    PRIMARY_PHONE_NUMBER STRING COMMENT 'Primary phone number for voice communications and emergency contact',
    SECONDARY_PHONE_NUMBER STRING COMMENT 'Alternative phone number for backup communication purposes',
    
    -- Company and Professional Information
    ORGANIZATION_NAME STRING COMMENT 'Official legal name of the customer organization or employing company',
    ORGANIZATION_DIVISION STRING COMMENT 'Specific business unit, department, or division within the organization',
    PROFESSIONAL_ROLE STRING COMMENT 'Job title or professional role of the individual within their organization',
    PROFESSIONAL_SENIORITY STRING COMMENT 'Level of seniority or hierarchy position (Executive, Manager, Individual Contributor)',
    DECISION_MAKING_AUTHORITY STRING COMMENT 'Level of purchasing or decision-making authority (Decision Maker, Influencer, User)',
    
    -- Address and Geographic Information
    MAILING_ADDRESS_LINE_1 STRING COMMENT 'Primary street address line for postal mail and shipping deliveries',
    MAILING_ADDRESS_LINE_2 STRING COMMENT 'Secondary address line for apartment, suite, or building details',
    MAILING_CITY STRING COMMENT 'City name for mailing address and geographic segmentation',
    MAILING_STATE_PROVINCE STRING COMMENT 'State, province, or region for mailing address and tax purposes',
    MAILING_POSTAL_CODE STRING COMMENT 'ZIP code or postal code for accurate mail delivery and location analytics',
    MAILING_COUNTRY_CODE STRING COMMENT 'ISO country code for international mailing and compliance requirements',
    TIME_ZONE STRING COMMENT 'Customer primary time zone for scheduling and communication timing',
    
    -- Business and Industry Classification
    INDUSTRY_SECTOR STRING COMMENT 'Primary industry sector or vertical market classification for segmentation',
    BUSINESS_SIZE_CATEGORY STRING COMMENT 'Organization size classification (SMB, Mid-Market, Enterprise, Large Enterprise)',
    ANNUAL_REVENUE_RANGE STRING COMMENT 'Estimated annual revenue range bracket for qualification and pricing',
    EMPLOYEE_COUNT_RANGE STRING COMMENT 'Estimated number of employees range for market segmentation',
    
    -- Customer Lifecycle and Engagement
    CUSTOMER_LIFECYCLE_STAGE STRING COMMENT 'Current position in customer journey (Prospect, Trial, Customer, Champion, At-Risk)',
    ENGAGEMENT_SCORE NUMBER COMMENT 'Calculated engagement score based on interactions across all touchpoints (0-100)',
    CUSTOMER_HEALTH_SCORE NUMBER COMMENT 'Overall customer health indicator combining usage, satisfaction, and risk factors (0-100)',
    PROPENSITY_TO_PURCHASE NUMBER COMMENT 'Likelihood score for making a purchase decision based on behavioral analysis (0-100)',
    CHURN_RISK_SCORE NUMBER COMMENT 'Calculated risk score for customer churn or cancellation (0-100)',
    
    -- Financial and Commercial Data
    TOTAL_LIFETIME_VALUE NUMBER COMMENT 'Calculated total lifetime value of customer relationship in USD',
    CURRENT_ANNUAL_CONTRACT_VALUE NUMBER COMMENT 'Current annual recurring revenue or contract value in USD',
    MONTHLY_RECURRING_REVENUE NUMBER COMMENT 'Monthly recurring revenue contribution in USD',
    PAYMENT_CREDITWORTHINESS STRING COMMENT 'Credit assessment category for payment terms and risk management',
    PREFERRED_PAYMENT_METHOD STRING COMMENT 'Customer preferred payment method for billing and invoicing',
    
    -- Communication and Preferences
    COMMUNICATION_PREFERENCE STRING COMMENT 'Preferred communication channel (Email, Phone, SMS, In-App, Portal)',
    MARKETING_OPT_IN_STATUS STRING COMMENT 'Marketing communication consent status for compliance and targeting',
    LANGUAGE_PREFERENCE STRING COMMENT 'Preferred language for communications and user interface',
    CONTENT_INTEREST_TOPICS STRING COMMENT 'Topics and content areas of interest for personalized marketing',
    
    -- System and Operational Data
    CUSTOMER_ACQUISITION_SOURCE STRING COMMENT 'Original source or channel through which customer was acquired',
    CUSTOMER_ACQUISITION_DATE DATE COMMENT 'Date when customer first became known to the organization',
    FIRST_PURCHASE_DATE DATE COMMENT 'Date of initial purchase or subscription activation',
    LAST_INTERACTION_DATE TIMESTAMP COMMENT 'Timestamp of most recent interaction across any system or channel',
    LAST_PURCHASE_DATE DATE COMMENT 'Date of most recent purchase or subscription renewal',
    
    -- Data Governance and Quality
    DATA_QUALITY_SCORE NUMBER COMMENT 'Calculated score indicating completeness and accuracy of customer data (0-100)',
    RECORD_LAST_UPDATED TIMESTAMP COMMENT 'Timestamp when customer record was last modified or synchronized',
    RECORD_CREATED_DATE TIMESTAMP COMMENT 'Timestamp when unified customer profile was first created',
    PRIMARY_DATA_SOURCE STRING COMMENT 'System identified as primary authoritative source for customer data',
    DATA_LINEAGE_SOURCES STRING COMMENT 'Comma-separated list of all systems contributing data to this profile'
)
COMMENT = 'Unified customer profile containing consolidated and enriched customer data from all source systems for 360-degree customer view'
TAG (SYSTEM_SOURCE = 'UNIFIED_CUSTOMER_DATA', DATA_CLASSIFICATION = 'MASTER_DATA',
     DATA_SENSITIVITY = 'COMPREHENSIVE_PII', BUSINESS_DOMAIN = 'CUSTOMER_360');
```

## 8. Data Quality and Validation Queries

Queries to validate the synthetic data and explore the schema structure


```python
-- View all tables created
SHOW TABLES IN DATABASE DATA_MAPPER_DB;

-- Check record counts in each source table
SELECT 
    'SF_CONTACTS_LEADS' as TABLE_NAME,
    COUNT(*) as RECORD_COUNT
FROM SOURCE_SYSTEMS.SF_CONTACTS_LEADS
UNION ALL
SELECT 
    'MARKETING_CONTACTS' as TABLE_NAME,
    COUNT(*) as RECORD_COUNT
FROM SOURCE_SYSTEMS.MARKETING_CONTACTS
UNION ALL
SELECT 
    'BILLING_CUSTOMERS' as TABLE_NAME,
    COUNT(*) as RECORD_COUNT
FROM SOURCE_SYSTEMS.BILLING_CUSTOMERS
UNION ALL
SELECT 
    'SUPPORT_CONTACTS' as TABLE_NAME,
    COUNT(*) as RECORD_COUNT
FROM SOURCE_SYSTEMS.SUPPORT_CONTACTS
UNION ALL
SELECT 
    'UNIFIED_CUSTOMER_PROFILE' as TABLE_NAME,
    COUNT(*) as RECORD_COUNT
FROM TARGET_SCHEMA.UNIFIED_CUSTOMER_PROFILE
ORDER BY TABLE_NAME;
```

```python
-- Sample data exploration - find common email addresses across systems
WITH email_analysis AS (
    SELECT 'SALESFORCE' as system, EMAIL_ADDRESS as email FROM SOURCE_SYSTEMS.SF_CONTACTS_LEADS
    UNION ALL
    SELECT 'MARKETING' as system, EMAIL_ADDR as email FROM SOURCE_SYSTEMS.MARKETING_CONTACTS
    UNION ALL
    SELECT 'BILLING' as system, BILLING_EMAIL as email FROM SOURCE_SYSTEMS.BILLING_CUSTOMERS
    UNION ALL
    SELECT 'SUPPORT' as system, CONTACT_EMAIL as email FROM SOURCE_SYSTEMS.SUPPORT_CONTACTS
)
SELECT 
    email,
    COUNT(DISTINCT system) as systems_count,
    LISTAGG(DISTINCT system, ', ') as systems_present
FROM email_analysis
GROUP BY email
ORDER BY systems_count DESC, email;
```
